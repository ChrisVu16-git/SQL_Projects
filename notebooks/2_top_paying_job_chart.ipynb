```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/Admin/Desktop/SQL_Projects/project_sql/2_top_paying_role.csv")

# Count unique jobs for each skill
skill_counts = (
    df.groupby("skills")["job_id"]
    .nunique()
    .sort_values(ascending=False)
)

# Calculate demand percentage
total_jobs = df["job_id"].nunique()
skill_percentage = skill_counts / total_jobs * 100

# Plot top 10 skills
skill_counts.head(10).sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Most Common Skills in Top-Paying DA Jobs")
plt.xlabel("Number of Job Postings")
plt.ylabel("Skill")
plt.tight_layout()
plt.show()
```
